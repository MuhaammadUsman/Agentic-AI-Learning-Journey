# **Categorical Data Visualization with Seaborn**

### 1. Aggregate Plots - Control Tendency Graphs
- Bar Plot 
- Count Plot 
### 2. Distribution Plot 
- Box Plot
- Violin Plot
- Swarm/Strip Plot
---------------------

# 1. Load Dataset

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset - health care
df = pd.read_csv("../datasets/healthcare.csv")
```

# 2. Count Plot

```python
# count plot - count treatment type
sns.countplot(data = df, x = "treatment_type", hue = "treatment_type")
plt.title("Number of Visits by Treatment Type")
plt.show()
```

# 3. Bar Plot

```python
sns.barplot(data = df, x = "department", y = "treatment_cost", hue = "department")
plt.xticks(rotation = 30)
plt.title("Average Treatment Cost by Department")
plt.show()
```

# 4. Box Plot
- shows median, quartiles and outliers

```python
# box plot
sns.boxplot(data = df, x = "department", y = "recovery_score", hue = "department")
plt.xticks(rotation = 30)
plt.show()
```

# 5. Violin Plot
- show full distribution shape

```python
# violin plot
sns.violinplot(data = df, x = "department", y = "recovery_score", hue = "department")
plt.xticks(rotation = 30)
plt.show()
```

# 6. Swarm Plot
- shows individual data points

```python
# swarm or strip plot
sns.stripplot(data = df, x = "department", y = "recovery_score", jitter = True, hue = "department")
plt.xticks(rotation = 30)
plt.show()
```

# 7. Comparison - Box plot and Strip Plot

```python
sns.boxplot(data = df, x = "department", y = "recovery_score", hue = "department")
sns.stripplot(data = df, x = "department", y = "recovery_score", color = "black", alpha = 0.7)
plt.xticks(rotation = 30)
plt.show()
```
