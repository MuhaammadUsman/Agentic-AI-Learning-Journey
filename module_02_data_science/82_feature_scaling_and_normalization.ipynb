# **Feature Scaling & Normalization**

### **Scaling Methods:**
- 1. Z-score Standardization
- 2. Min-Max Scaling
### **Data Leakage Risk:**
- Applying scaling on both training and test data together leads to data leakage, as test data information is used during training.
### **When & Why to Scale:**
- Scale features when using distance-based models (KNN, SVM, PCA, Clustering).
- Avoid or handle carefully for tree-based models (Decision Trees, Random Forest), as they are insensitive to scale.

# **1. Z-score Scaling Standardization Manual**
## 1.1: Calculate Mean & STD

```python
income_mean = df["monthly_income"].mean()
income_std = df["monthly_income"].std()

print("Mean:", income_mean)
print("STD:", income_std)
```

## 1.2: Apply Z-score Formula

```python
df["income_zscore"] = (df["monthly_income"] - income_mean) / income_std

# display monthly_income and income_zscore
df[["monthly_income", "income_zscore"]].head() 
```

```python
# Verify Z-score scaling
df["income_zscore"].mean(), df["income_zscore"].std()
```

# **2. Min-Max Scaling (map values: 0 to 1)**
## 2.1: Calculate Min & Max

```python
bill_min = df["monthly_bill"].min()
bill_max = df["monthly_bill"].max()

print("Min:", bill_min)
print("Max:", bill_max)
```

## 2.2: Apply Min-Max Formula

```python
df["bill_min_max"] = (df["monthly_bill"] - bill_min) / (bill_max - bill_min)

# display monthy_bill & bill_min_max
df[["monthly_bill", "bill_min_max"]].head()
```

```python
# Verify Min-Max Scaling
df["bill_min_max"].min(), df["bill_min_max"].max()
```

```python
df.loc[
    [df["monthly_bill"].idxmin(),
     df["monthly_bill"].idxmax(),
     df["monthly_bill"].sort_values().index[len(df)//2]],
    ["monthly_bill", "bill_min_max"]]
```
