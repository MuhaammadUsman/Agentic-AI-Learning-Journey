# **Creating New Features from Existing Data**

# 1. Load Dataset

```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# load dataset - telecom churn dataset
df = pd.read_csv("../datasets/telecom_churn.csv")
# display dataset
df.head()
```

# 2. Extract Data Part - signup_data col

```python
# convert date column to datetime format
df["signup_date"] = pd.to_datetime(df["signup_date"])
```

```python
# extract year, month, and day from signup_date column
df["signup_year"] = df["signup_date"].dt.year
df["signup_month"] = df["signup_date"].dt.month
df["signup_day"] = df["signup_date"].dt.dayofweek

# show new features
df[["signup_year", "signup_month", "signup_day"]].head()
```
