# **Creating New Features from Existing Data**

# 1. Load Dataset

```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# load dataset - telecom churn dataset
df = pd.read_csv("../datasets/telecom_churn.csv")
# display dataset
df.head()
```

# 2. Extract Data Part - signup_data col

```python
# convert date column to datetime format
df["signup_date"] = pd.to_datetime(df["signup_date"])
```

```python
# extract year, month, and day from signup_date column
df["signup_year"] = df["signup_date"].dt.year
df["signup_month"] = df["signup_date"].dt.month
df["signup_day"] = df["signup_date"].dt.dayofweek

# show new features
df[["signup_year", "signup_month", "signup_day"]].head()
```

# 3. Ratio Feature Creation

```python
# bill_income ratio
df["bill_income_ratio"] = df["monthly_bill"] / df["monthly_income"]

# show new feature
df[["monthly_bill", "monthly_income", "bill_income_ratio"]].head() 

# higher ratio, higher financial stress, higher churn risk
```

# 4. Interaction Features

```python
# Interaction features are created by combining two or more existing features
# higher interaction value, higher churn risk
df ["support_bill_interaction"] = df["support_tickets"] * df["monthly_bill"]

# show new feature
df[["support_tickets", "monthly_bill", "support_bill_interaction"]].head()
```

# 5. Text Length Feature

```python
# text length features
# longer feedback-more detailed feedback-more likely to churn
df["feedback_length"] = df["customer_feedback"].str.len()

# show new feature
df[["customer_feedback", "feedback_length"]].head()
```

# 6. Keyward Flag Features

```python
# binary flag makes text ML friendly
df["complaint_flag"] = df["customer_feedback"].str.contains("slow|bad|poor|expensive|unreliable|problem|issue|drops",
                                                            case = False).astype(int)

# show new feature
df[["customer_feedback", "complaint_flag"]].head()
```

# 7. Final Features Review

```python
# display all new features
df[["signup_year", "signup_month", "signup_day", "bill_income_ratio", "support_bill_interaction", "feedback_length", "complaint_flag"]].head()
```
