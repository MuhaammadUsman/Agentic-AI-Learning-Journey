# **Introduction to Seaborn Visualization**

# 1. Import Libraries & Dataset

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset - healthcare dataset
df = pd.read_csv("../datasets/healthcare.csv")
```

```python
# sample size and display 1st five rows
df = df.sample(500)
df.head()
```

# 2. Scatter Plot (Axes Level)

```python
# scatter plot
sns.scatterplot(data = df, x = "treatment_cost", y = "recovery_score", hue = "visit_type")
plt.title("Treatment Cost vs Recovery Score")
plt.show()
```

# 3.  Relational Plot - Figure Level

```python
# set theme
sns.set_theme(style = "darkgrid")
# relplot
sns.relplot(data = df, x = "treatment_cost", y = "recovery_score", hue = "visit_type", col = "gender")
plt.show()
```
