# **Customizing & Exporting Matplotlib Visualizations**

# 1. Import Libraries and Load Dataset

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load dataset - ecommerce sales
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])
```

```python
# count total number of orders for each payment method
payment_counts = df["payment_method"].value_counts()

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    payment_counts.index,
    payment_counts.values,
    color=["skyblue", "orange", "lightgreen"],
    edgecolor="black"
)

# Labels and title
ax.set_title("Payment Method Distribution")
ax.set_xlabel("Payment Method")
ax.set_ylabel("Number of Orders")

# Tick customization
ax.tick_params(axis = "x", rotation = 45, labelsize = 12)

# Add legend for each payment method
ax.legend(bars, payment_counts.index, title = "Payment Method")

plt.tight_layout()
plt.show()
```
