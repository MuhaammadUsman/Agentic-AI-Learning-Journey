# **Encoding Categorical Variables**
- Data science tools work only with numeric data
- Convert text data into numeric form
- Essential before training a model

#### **Methods:** 
- 1. One-hot Encoding
- 2. Ordinal Encoding
--------------------


# 1. Import Libraries

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# create dataframe - sample data
df = pd.DataFrame({
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Segment": ["A", "B", "C", "B", "A"],
    "Sale": [200, 300, 500, 400, 120]
    })
print(df)
```

# 2. One-hot Encoding

```python
# one-hot encoding
encoded = pd.get_dummies(df, columns = ["City", "Segment"], drop_first = True)
print(encoded)
```

# 2. Ordinal Encoding

```python
# import sklearn 
from sklearn.preprocessing import OrdinalEncoder
```

```python
# sample data
df2 = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "Medium", "Large"]
})
print(df2)
```

```python
# encoding
encoder = OrdinalEncoder(categories = [["Small", "Medium", "Large"]])
df2["size_encoded"] = encoder.fit_transform(df2[["Size"]])

print(df2)
```
